## 1. My lane as an ML task (type)

Lane: Content refresh / declining-page prioritization

ML task type: Ranking

The goal is to rank content pages by how likely they are to need a refresh. Instead of treating every page equally, the model should place the pages most likely to be declining near the top so that a content team can review the highest-priority pages first.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Number of pages available for prioritization:", len(df))
print("\nExample of pages ordered by impressions:")

df[["content_id", "impressions_90d"]].sort_values(
    "impressions_90d", ascending=False
).head(10)
```

## 2. Target or proxy

Target: is_declining_label

For the starter dataset, I define the target as whether a page's recent trend is declining: trend_direction == "down".

This converts the business question into a measurable ML target. The model would use information available before the decision to estimate which pages are more likely to be declining.

For a future production system, the target could be refined using a future performance window so that the prediction is made before the outcome is observed.

```python
df["is_declining_label"] = (
    df["trend_direction"].str.lower() == "down"
).astype(int)

print("Declining pages:", df["is_declining_label"].sum())
print("Total pages:", len(df))
print(
    "Declining rate:",
    round(df["is_declining_label"].mean(), 3)
)
```

## 3. Success metric

Primary metric: Precision@50

Precision@50 measures the proportion of the top 50 ranked pages that are actually declining.

This metric matches the real action because a content team has limited review capacity. A useful system should put a high proportion of genuinely declining pages near the top of the review queue rather than simply achieving high overall accuracy across all pages.

```python
top_50 = df.sort_values(
    "impressions_90d", ascending=False
).head(50)

precision_at_50 = top_50["is_declining_label"].mean()

print("Precision@50 for this example ranking:", round(precision_at_50, 3))
print(
    "Number of declining pages in the top 50:",
    top_50["is_declining_label"].sum()
)
```

## 4. The unit of analysis, as a real dataframe

One row = one content page/search-content record.

Each row contains observable characteristics of that page, such as its age, time since its last update, search impressions, average position, CTR, and recent trend direction.

The model would use the observable page-level information to prioritize which individual pages should be reviewed for a potential content refresh.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

lane_df = df[
    [
        "content_id",
        "content_type",
        "content_age_days",
        "days_since_last_update",
        "impressions_90d",
        "avg_position",
        "ctr",
        "trend_direction"
    ]
].copy()

print("Rows:", len(lane_df))
print("One row represents one content page/search-content record.")

lane_df.head(10)
```

## 5. Why ML beats a fixed rule here

A fixed rule such as "prioritize pages that are old and have many impressions" is easy to understand, but it relies on manually chosen thresholds and only captures the relationships we explicitly specify.

ML can evaluate several page-level signals together and learn combinations of signals associated with declining pages. This can help produce a ranked review queue instead of relying on one manually designed rule.

The purpose is not to assume that the model is always better. Its value should be tested against a transparent baseline using an appropriate validation strategy.

```python
candidate_features = [
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "avg_position",
    "ctr",
    "word_count"
]

print("Candidate signals available to the model:")
for feature in candidate_features:
    print("-", feature)

print("\nNumber of candidate signals:", len(candidate_features))
```
